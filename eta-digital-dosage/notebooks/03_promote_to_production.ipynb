# Promote a validated MLflow model version

```python
import os
from mlflow import MlflowClient
MODEL_NAME = os.getenv('MLFLOW_REGISTERED_MODEL_NAME', 'eta-digital-dosage-predictor')
client = MlflowClient()
```

```python
versions = client.search_model_versions(f"name='{MODEL_NAME}'")
if not versions:
    raise RuntimeError(f'No registered versions found for {MODEL_NAME}')
latest = max(versions, key=lambda item: int(item.version))
client.set_registered_model_alias(MODEL_NAME, 'champion', latest.version)
{'model': MODEL_NAME, 'alias': 'champion', 'version': latest.version}
```

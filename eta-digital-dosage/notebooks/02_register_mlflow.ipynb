# Register the prediction model in MLflow

```python
from pathlib import Path
import os, pickle, sys
import mlflow
PROJECT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(PROJECT / 'src'))
from eta_digital.mlflow import EtaDigitalPredictionModel, prediction_input_example, register_prediction_model
```

```python
mlflow.set_tracking_uri(os.getenv('MLFLOW_TRACKING_URI', 'http://localhost:5000'))
mlflow.set_experiment(os.getenv('MLFLOW_EXPERIMENT_NAME', 'eta-digital-dosage'))
with (PROJECT / 'artifacts/model.pkl').open('rb') as handle:
    model = pickle.load(handle)
wrapped = EtaDigitalPredictionModel(model)
with mlflow.start_run() as run:
    info = register_prediction_model(wrapped, 'prediction_model', os.getenv('MLFLOW_REGISTERED_MODEL_NAME', 'eta-digital-dosage-predictor'), prediction_input_example())
(run.info.run_id, info.model_uri)
```
